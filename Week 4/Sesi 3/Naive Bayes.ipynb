Setelah fitur dan target dipersiapkan, dataset dibagi menjadi data training dan data testing menggunakan train_test_split. Parameter test_size=0.2 berarti 20% data digunakan sebagai data testing, sedangkan 80% sisanya digunakan sebagai data training. Data training digunakan untuk membangun dan melatih model, sedangkan data testing digunakan untuk mengetahui kemampuan model dalam melakukan prediksi terhadap data yang tidak digunakan selama proses pelatihan. Parameter random_state=0 digunakan agar pembagian data tetap konsisten ketika program dijalankan kembali.

```python
from sklearn.naive_bayes import GaussianNB

model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

Penjelasan 

Sebelum membuat model, modul menjelaskan bahwa terdapat tiga jenis utama Naïve Bayes, yaitu Gaussian Naïve Bayes, Multinomial Naïve Bayes, dan Bernoulli Naïve Bayes. Gaussian Naïve Bayes digunakan untuk fitur numerik kontinu yang diasumsikan mengikuti distribusi Gaussian atau normal. Multinomial Naïve Bayes digunakan untuk data yang berbasis frekuensi kejadian, sedangkan Bernoulli Naïve Bayes digunakan untuk data biner atau data dengan nilai 0 dan 1. Karena seluruh fitur pada dataset Iris berupa data numerik kontinu, modul memilih Gaussian Naïve Bayes untuk proses klasifikasi.

Warning tersebut muncul karena model Gaussian Naïve Bayes dilatih menggunakan X_train yang berasal dari DataFrame, sehingga model mengenali nama-nama fitur seperti sepal length (cm), sepal width (cm), petal length (cm), dan petal width (cm).
```python
#prediksi inputan data baru
#input_data = pd.DataFrame(np. array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)

input_data = pd.DataFrame(
    np.array([[5.1, 3.5, 1.4, 0.2]]),
    columns=iris.feature_names
)

result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)

print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

Penjelasan

Nama kolom diberikan menggunakan iris.feature_names, sehingga struktur data baru menjadi sama dengan struktur fitur yang digunakan ketika model Gaussian Naïve Bayes dilatih. Setelah itu data diprediksi menggunakan NB_model.predict(), kemudian hasil numeriknya dikembalikan menjadi nama spesies menggunakan lab.inverse_transform().
