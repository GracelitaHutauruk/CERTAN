```python
def f(x,y):
 return np.cos(np.exp(-x**2-y**2))*np.sin(np.sqrt(x**2+y**2))
x = np.linspace(-np.pi, np.pi, 30)
y = np.linspace(-np.pi, np.pi, 30)
X,Y = np.meshgrid(x,y)
Z = f(X,Y)
fig = plt.figure()
ax = plt.axes(projection = '3d')
ax.contour3D(X, Y, Z, 50, cmap = 'binary')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Membuat grid titik untuk sumbu x dan y
x = np.linspace(-1.5, 1.5, 400)
y = np.linspace(-1.5, 1.5, 400)
X, Y = np.meshgrid(x, y)

# Rumus persamaan hati 2 dimensi sesuai modul tugas
Z = X**6 + 3*X**4*Y**2 - 3*X**4 + 3*X**2*Y**4 - X**2*Y**3 - 6*X**2*Y**2 + 3*X**2 + Y**6 - 3*Y**4 + 3*Y**2

# Menggambar garis hati pada nilai 1
plt.contour(X, Y, Z, [1], colors='red')
plt.title("Grafik Hati 2D")
plt.xlabel("Sumbu X")
plt.ylabel("Sumbu Y")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

fig = plt.figure(figsize=(8, 6))
ax = plt.axes(projection='3d')

# Grid koordinat X dan Y
x = np.linspace(-2, 2, 200)
y = np.linspace(-2, 2, 200)
X, Y = np.meshgrid(x, y)

# Persamaan z sesuai soal nomor 3: z = (5y/4 - |x|)^2 + x^2
Z = ((5 * Y / 4) - np.abs(X)) ** 2 + X**2

# Plot permukaan 3D
surf = ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none', alpha=0.8)

ax.set_title(r'Grafik 3D: $z = \left(\frac{5y}{4} - |x|\right)^2 + x^2$')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')

fig.colorbar(surf, shrink=0.5, aspect=5)
plt.show()
```
