```python
import pandas as pd

# Membaca dataset Global Weather Repository
path = "GlobalWeatherRepository.csv"
df = pd.read_csv(path)
df
```

```python
# Analisis korelasi
correlation = df[
    ['temperature_fahrenheit',
     'pressure_mb',
     'air_quality_Sulphur_dioxide']
].corr()

correlation
```

```python
print("Korelasi Temperature Fahrenheit, Pressure, dan Sulphur Dioxide:")
print(correlation)
```
